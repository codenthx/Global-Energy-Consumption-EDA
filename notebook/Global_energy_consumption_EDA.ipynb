## Dataset Exploration

The dataset was explored using Pandas functions such as `info()`, `describe()`, `isnull()`, and `duplicated()`.

These methods helped identify:
- Data types
- Missing values
- Duplicate records
- Overall statistical characteristics

This exploration is an important step before cleaning and visualizing the dataset.

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
# Remove columns where more than 50% of the values are missing

threshold = len(df) * 0.5
df = df.dropna(axis=1, thresh=threshold)

print("Columns with excessive missing values removed.")
```

```python
# Fill missing numerical values with the median

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

print("Missing numerical values filled successfully!")
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(12,6))

df.groupby('year')['primary_energy_consumption'].mean().plot()

plt.title("Average Primary Energy Consumption Over Time")
plt.xlabel("Year")
plt.ylabel("Average Energy Consumption")

plt.savefig("graph1.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

The visualization illustrates how average primary energy consumption has changed over time. This helps identify long-term trends in global energy usage.

```python
top = df.groupby('country')['primary_energy_consumption'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))

top.plot(kind='bar')

plt.title("Top 10 Countries by Average Energy Consumption")

plt.ylabel("Energy Consumption")

plt.savefig("graph2.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

The chart compares the average energy consumption of the top ten countries, highlighting those with the highest demand.

```python
# ==========================================
# Remove Duplicate Rows
# ==========================================

duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)

df = df.drop_duplicates()

print("Dataset Shape after removing duplicates:", df.shape)
```

```python
df.isnull().sum()
```

```python
# Fill missing numerical values using median

numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

print("Numerical missing values filled!")
```

```python
# Fill missing categorical values

categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Categorical missing values filled!")
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(12,6))

df['year'].hist(bins=30)

plt.title("Distribution of Records by Year")
plt.xlabel("Year")
plt.ylabel("Count")

plt.savefig("graph3.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

This histogram shows how many records exist for each year in the dataset. It helps identify periods where more energy data was collected.

```python
plt.figure(figsize=(12,6))

df['country'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 Countries in Dataset")
plt.xlabel("Country")
plt.ylabel("Number of Records")

plt.xticks(rotation=45)

plt.savefig("graph4.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

The graph shows which countries contribute the most records to the dataset. Countries with more observations often have richer historical energy data.

```python
plt.figure(figsize=(10,6))

df.boxplot(column='population')

plt.title("Boxplot of Population")
plt.ylabel("Population")

plt.savefig("graph4_boxplot.png", dpi=300, bbox_inches="tight")

plt.show()
```

### Observation

The boxplot highlights the distribution of population values and reveals the presence of outliers. These outliers correspond to countries with exceptionally large populations, such as India and China, and are expected in a global dataset.

```python
import seaborn as sns
```

```python
plt.figure(figsize=(14,10))

corr = df.select_dtypes(include='number').corr()

sns.heatmap(corr,
            cmap="coolwarm",
            linewidths=0.5)

plt.title("Correlation Heatmap")

plt.savefig("graph5_heatmap.png", dpi=300, bbox_inches="tight")

plt.show()
```

### Observation

The correlation heatmap illustrates relationships between numerical variables. Strong positive correlations indicate variables that increase together, while negative correlations show inverse relationships. This helps identify features that may be useful for future predictive modeling.

```python
plt.figure(figsize=(12,6))

df.groupby('year')['population'].mean().plot()

plt.title("Average Population Over Time")
plt.xlabel("Year")
plt.ylabel("Average Population")

plt.savefig("graph6_population_trend.png", dpi=300, bbox_inches="tight")

plt.show()
```

### Observation

The average population shows an increasing trend over time, reflecting global population growth captured within the dataset.

```python
top_pop = df.groupby('country')['population'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))

top_pop.plot(kind='bar')

plt.title("Top 10 Countries by Average Population")
plt.xlabel("Country")
plt.ylabel("Average Population")

plt.xticks(rotation=45)

plt.savefig("graph7_top_population.png", dpi=300, bbox_inches="tight")

plt.show()
```

### Observation

China and India are expected to appear among the countries with the highest average population, demonstrating their significant contribution to global population statistics.

```python
countries_per_year = df.groupby('year')['country'].nunique()

plt.figure(figsize=(12,6))

countries_per_year.plot()

plt.title("Countries Recorded Per Year")

plt.xlabel("Year")

plt.ylabel("Number of Countries")

plt.savefig("graph8_countries_per_year.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

The dataset coverage increases over time, showing that more countries contribute energy statistics in recent years.

```python
plt.figure(figsize=(12,6))

plt.hist(df['population'], bins=30)

plt.title("Population Distribution")
plt.xlabel("Population")
plt.ylabel("Frequency")

plt.savefig("graph9_population_distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```

### 📌 Observation

The histogram shows the distribution of population values in the dataset. Most records represent countries with relatively smaller populations, while a few countries have exceptionally large populations, resulting in a right-skewed distribution.
